# Classificar os ativos

Esse notebook tem como objetivo adicionar uma coluna de classificação de ativos para cada um dos investimentos feitos.

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('../data/raw/PosicaoIntelli.csv')
df.head()
```

Note como o dataframe possui uma coluna "TIPO DE ATIVO", que agrupa os ativos em grupos que possuem o mesmo nível de risco. O próximo passo é descobrir quais são todos esses tipos e classificar eles manualmente.

```python
tipos_unicos = df['TIPO DE ATIVO'].unique()
print(tipos_unicos)
```

Uma vez descoberto quais são todos os tipos de ativos existentes nos dados, utilizando o auxilio da PerplexityAI, classificamos cada um dos tipos de fundos em Conservador, Moderado ou Sofisticado. Assim, criamos um csv chamado "Barema".

```python
barema = pd.read_csv('../data/raw/classificacao_de_tipos.csv')
barema.tail()
```

Uma vez com um barema definido, basta alocarmos as respectivas classificações para cada tipo de ativo nas linhas do dataframe original.

```python
df['TIPO DE ATIVO'] = df['TIPO DE ATIVO'].str.strip()
barema['Ativo'] = barema['Ativo'].str.strip()

mapeamento_dict = dict(zip(barema['Ativo'], barema['Classificacao']))

df['CLASSIFICACAO DO ATIVO'] = df['TIPO DE ATIVO'].map(mapeamento_dict)
df.head()
```

Agora com a coluna Classificação do Ativo (classificada de acordo com o Tipo de Ativo), basta apenas verificar se todos os ativos estão de fato com uma classificação

```python
print(df['CLASSIFICACAO DO ATIVO'].value_counts())


print("\nQuantidade de valores nulos em CLASSIFICACAO DO ATIVO:")
print(df['CLASSIFICACAO DO ATIVO'].isna().sum())


print("\nExemplos de registros com classificação nula:")
print(df[df['CLASSIFICACAO DO ATIVO'].isna()].head())


missing_mappings = set(df[df['CLASSIFICACAO DO ATIVO'].isna()]['TIPO DE ATIVO'].unique())
print("\nTipos de ativos sem classificação no barema:")
for asset_type in missing_mappings:
    print(f"- {asset_type}")
```

Como é possível observar, nenhum valor de classificação do ativo foi dado como nulo, logo, todos os ativos foram perfeitamente classificados
